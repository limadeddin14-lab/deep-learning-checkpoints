**1. Conceptual Understanding**

A GPU is more efficient than a CPU for many deep learning tasks because it can perform a large number of calculations in parallel. GPUs are especially effective for operations such as matrix multiplication and other parallel numerical computations commonly used during neural network training.

**3. Reflection Question**

Using a GPU would be helpful when training a model on 10,000 images because image-based deep learning involves a large number of numerical and matrix operations. The GPU can process many of these operations in parallel, reducing training time compared with relying only on the CPU.
